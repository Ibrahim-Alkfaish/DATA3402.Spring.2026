*Exercise 4:* Write a function that takes a `n` by `n` matrix representing a tic-tac-toe game, and returns -1, 0, 1, or 2 indicating the game is incomplete, the game is a draw, player 1 has won, or player 2 has one, respectively. Here are some example inputs you can use to test your code:

````python
winner_is_2 = [[2, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_1 = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 1]]

winner_is_also_1 = [[0, 1, 0],
	[2, 1, 0],
	[2, 1, 1]]

no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 2]]

also_no_winner = [[1, 2, 0],
	[2, 1, 0],
	[2, 1, 0]]
````

````python
def round_one(board):
    n = len(board)
    for i in range(n):
        if board[i][0] != 0:
            win = True
            for j in range(n):
                if board[i][j] != board[i][0]:
                    win = False
            if win:
                return board[i][0]

    for j in range(n):
        if board[0][j] != 0:
            win = True
            for i in range(n):
                if board[i][j] != board[0][j]:
                    win = False
            if win:
                return board[0][j]

    if board[0][0] != 0:
        win = True
        for i in range(n):
            if board[i][i] != board[0][0]:
                win = False
        if win:
            return board[0][0]

    if board[0][n-1] != 0:
        win = True
        for i in range(n):
            if board[i][n-1-i] != board[0][n-1]:
                win = False
        if win:
            return board[0][n-1]

    for i in range(n):
        for j in range(n):
            if board[i][j] == 0:
                return -1

    return 0
````

*Exercise 9:* Use all of the previous exercises to implement a full tic-tac-toe game, where an appropriate board is drawn, 2 players are repeatedly asked for a location coordinates of where they wish to place a mark, and the game status is checked until a player wins or a draw occurs.

````python
def full_game(n,m):
    board = make_board(n)
    player = 1
    
    while True:
        tic_tac_toe(board)
        
        print("Player", player, "enter your move as row and column numbers (1-" + str(n) + ")")
        move = input("Example: '2 3' for row 2, column 3: ")
        
        parts = move.split()
        if len(parts) != 2:
            print("Invalid input, try again.")
            continue
        
        try:
            x = int(parts[0]) - 1
            y = int(parts[1]) - 1
        except:
            print("Invalid numbers, try again.")
            continue
        
        if x < 0 or x >= n or y < 0 or y >= n:
            print("Move out of bounds, try again.")
            continue
        
        if not place_move(board, player, x, y):
            print("Cell already taken, try again.")
            continue
        
        status = round_one(board)
        if status == 1:
            tic_tac_toe(board)
            print("Player 1 wins!")
            break
        elif status == 2:
            tic_tac_toe(board)
            print("Player 2 wins!")
            break
        elif status == 0:
            tic_tac_toe(board)
            print("It's a draw!")
            break
        
        if player == 1:
            player = 2
        else:
            player = 1
````

*Exercise 11:* Develop a version of the game where one player is the computer. Do not do an extensive search for the best move. Make computer player can simply protect against loosing and otherwise try to win with straight or diagonal patterns.

````python
# Write you solution here
def get_lines():
    return [
        # rows
        [(0,0),(1,0),(2,0)],
        [(0,1),(1,1),(2,1)],
        [(0,2),(1,2),(2,2)],

        # columns
        [(0,0),(0,1),(0,2)],
        [(1,0),(1,1),(1,2)],
        [(2,0),(2,1),(2,2)],

        # diagonals
        [(0,0),(1,1),(2,2)],
        [(2,0),(1,1),(0,2)]
    ]
````

````python
def line_move(board, player):
    for line in get_lines():
        values = [board[y][x] for (x, y) in line]

        if values.count(player) == 2 and values.count(0) == 1:
            empty_index = values.index(0)
            return line[empty_index]

    return None
````

````python
def computer_move(board):
    move = line_move(board, 2)
    if move:
        place_move(board, 2, move[0], move[1])
        return
    move = line_move(board, 1)
    if move:
        place_move(board, 2, move[0], move[1])
        return

    if board[1][1] == 0:
        place_move(board, 2, 1, 1)
        return

    for y in range(3):
        for x in range(3):
            if board[y][x] == 0:
                place_move(board, 2, x, y)
                return
````

*Exercise 12:* Develop a version of the game where one player is the computer. This time, write a computer player using exhaustive search with a max depth parameter, similar to lecture.

````python
# Write you solution here
def copy_board(board):
    return [row[:] for row in board]
````

````python
def minimax(board, depth, max_depth, player):
    result = round_one(board)

    # terminal states
    if result == 2:   # computer wins
        return 10 - depth
    if result == 1:   # human wins
        return depth - 10
    if result == 0:   # draw
        return 0
    if depth == max_depth:
        return 0

    if player == 2:  # maximize
        best = -1000
        for r in range(3):
            for c in range(3):
                if board[r][c] == 0:
                    new_board = copy_board(board)
                    new_board[r][c] = 2
                    score = minimax(new_board, depth + 1, max_depth, 1)
                    best = max(best, score)
        return best
    else:  # minimize
        best = 1000
        for r in range(3):
            for c in range(3):
                if board[r][c] == 0:
                    new_board = copy_board(board)
                    new_board[r][c] = 1
                    score = minimax(new_board, depth + 1, max_depth, 2)
                    best = min(best, score)
        return best
````

````python
def computer_move_exhaustive(board, max_depth):
    best_score = -1000
    best_move = None

    for r in range(3):
        for c in range(3):
            if board[r][c] == 0:
                new_board = copy_board(board)
                new_board[r][c] = 2
                score = minimax(new_board, 0, max_depth, 1)

                if score > best_score:
                    best_score = score
                    best_move = (r, c)

    if best_move:
        place_move(board, 2, best_move[0], best_move[1])
````

````python
def full_game_vs_computer(max_depth):
    board = make_board(3)

    while True:
        tic_tac_toe(board)

        # human move
        move = input("Your move (row col): ")
        r, c = map(int, move.split())
        r -= 1
        c -= 1

        if not place_move(board, 1, r, c):
            print("Invalid move.")
            continue

        status = round_one(board)
        if status != -1:
            break

        # computer move
        computer_move_exhaustive(board, max_depth)

        status = round_one(board)
        if status != -1:
            break

    tic_tac_toe(board)
    if status == 1:
        print("You win!")
    elif status == 2:
        print("Computer wins!")
    else:
        print("Draw!")
````

````python
def computer_move_minimax(board, depth):
    best_score = -1000
    best_move = None

    for y in range(3):
        for x in range(3):
            if board[y][x] == 0:
                board[y][x] = 2
                score = minimax(board, depth - 1, False)
                board[y][x] = 0

                if score > best_score:
                    best_score = score
                    best_move = (x, y)

    if best_move:
        place_move(board, 2, best_move[0], best_move[1])
````

*Exercise 13:* Make the 2 computer players play each-other for 10 games on a 3x3, then 4x4, then 5x5 grid. Set the max depth so that the games only take seconds. Measure the "smarter" player's win rate for each grid.

````python
def computer_move_two(board, player, max_depth):
    n = len(board)

    best_score = -1000 if player == 2 else 1000
    best_move = None

    for r in range(n):
        for c in range(n):
            if board[r][c] == 0:
                new_board = copy_board(board)
                new_board[r][c] = player
                score = minimax(
                    new_board, 0, max_depth,
                    1 if player == 2 else 2
                )

                if player == 2 and score > best_score:
                    best_score = score
                    best_move = (r, c)
                if player == 1 and score < best_score:
                    best_score = score
                    best_move = (r, c)

    if best_move:
        board[best_move[0]][best_move[1]] = player
````

````python
def play_game(n, depth_smart, depth_dumb, smart_player):
    board = make_board(n)
    player = 1

    while True:
        if player == smart_player:
            computer_move_two(board, player, depth_smart)
        else:
            computer_move_two(board, player, depth_dumb)

        status = round_one(board)
        if status != -1:
            return status

        player = 2 if player == 1 else 1
````

````python
def run_experiment(n, depth_smart, depth_dumb):
    smart_wins = 0
    games = 10

    for i in range(games):
        smart_player = 1 if i % 2 == 0 else 2
        winner = play_game(n, depth_smart, depth_dumb, smart_player)

        if winner == smart_player:
            smart_wins += 1

    win_rate = smart_wins / games
    print(f"{n}x{n} grid: smart win rate = {win_rate}")
````

````python
# Test your solution here
run_experiment(3, depth_smart=6, depth_dumb=2)
run_experiment(4, depth_smart=3, depth_dumb=1)
run_experiment(5, depth_smart=2, depth_dumb=1)
````
